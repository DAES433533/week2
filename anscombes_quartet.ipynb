# Complexity of Statistics

## Anscombe's Quartet

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
file = 'anscombe.txt'
df = pd.read_csv(file)
df
```

### Plot Anscombe's Data

```python
fig, ax = plt.subplots(
    2, 2,
    figsize=(11,6),
    sharex=True,
    sharey=True,
    gridspec_kw={'wspace': 0, 'hspace': 0}
)

datasets = ['I', 'II', 'III', 'IV']

for a, ds in zip(ax.flat, datasets):
    data = df[df['dataset'] == ds]

    a.scatter(data.x, data.y, label = ds)
    a.set_xlim(2, 19.5)
    a.set_ylim(2, 15)

    a.text(2.5, 13.5, ds, fontsize=12)
```

### Use Pandas to determine statistics

```python
df_grouped = df.groupby('dataset')[['x', 'y']]
```

```python
# Mean
df_grouped.mean()
```

```python
# Variance
df_grouped.var()
```

```python
# X-Y Correlation
print('dataset\tcorr\tR-squared')
for ds, data in df_grouped:
    corr = data['x'].corr(data['y'])
    r2 = corr**2 
    print(f"{ds}\t{corr:.4f}\t{r2:.4f}")
```

```python
# Regression and R-squared 
def regression(data):

    # Calculate the slope of the best-fit line
    # slope = covariance(x, y) / variance(x)
    slope = data['x'].cov(data['y']) / data['x'].var()

    # Calculate the y-intercept
    # intercept = mean(y) - slope * mean(x)
    intercept = data['y'].mean() - slope * data['x'].mean()

    # Return the calculated values as a pandas Series.
    return pd.Series({
        'slope': slope,
        'intercept': intercept
    })

df_regress = df_grouped.apply(regression)
df_regress
```

### Plot the best fit lines

```python
fig, ax = plt.subplots(
    2, 2,
    figsize=(11,6),
    sharex=True,
    sharey=True,
    gridspec_kw={'wspace': 0, 'hspace': 0}
)

datasets = ['I', 'II', 'III', 'IV']
x_fit = np.arange(0, 20, 1)

for a, ds in zip(ax.flat, datasets):
    data = df[df['dataset'] == ds]
    best_fit = df_regress.loc[ds] # 'dataset' is now the DataFrame index
    y_fit = best_fit.intercept + best_fit.slope * x_fit

    a.scatter(data.x, data.y, label = ds)
    a.plot(x_fit, y_fit, '--')

    a.set_xlim(2, 19.5)
    a.set_ylim(2, 15)

    a.text(2.5, 13.5, ds, fontsize=12)
```
